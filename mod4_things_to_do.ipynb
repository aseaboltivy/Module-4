```python
import csv
import sqlite3
import os

# 1. Ensure books2.csv exists
if not os.path.exists('books2.csv'):
    with open('books2.csv', 'w', encoding='utf-8') as f:
        f.write("title,author,year\n")
        f.write("The Weirdstone of Brisingamen,Alan Garner,1960\n")
        f.write("Perdido Street Station,China Miéville,2000\n")
        f.write("Thud!,Terry Pratchett,2005\n")
        f.write("The Spellman Files,Lisa Lutz,2007\n")
        f.write("Small Gods,Terry Pratchett,1992\n")

# 2. Connect, create the table (fixes the error!), and insert data
conn = sqlite3.connect('books.db')
curs = conn.cursor()

# Drop table if it exists to avoid duplicate primary key/row conflicts on re-runs
curs.execute('DROP TABLE IF EXISTS books')

curs.execute('''
    CREATE TABLE books (
        title TEXT,
        author TEXT,
        year INTEGER
    )
''')

ins = 'INSERT INTO books (title, author, year) VALUES (?, ?, ?)'

with open('books2.csv', 'rt', encoding='utf-8') as infile:
    reader = csv.DictReader(infile)
    for row in reader:
        clean_row = {k.strip(): v.strip() for k, v in row.items()}
        curs.execute(ins, (clean_row['title'], clean_row['author'], int(clean_row['year'])))

conn.commit()
conn.close()

print("Table 'books' created and data inserted successfully!")
```
